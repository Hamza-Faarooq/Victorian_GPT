```python
import re
import os

raw = "VictorianGPT/raw"
clean = "VictorianGPT/cleaned"

# 1. Forge the Directory (Prevents FileNotFoundError)
os.makedirs(clean, exist_ok=True)

print("Commencing the cleansing of the grand archives...\n")

for file in os.listdir(raw):
    # Ignore hidden system files like .DS_Store
    if not file.endswith(".txt"):
        continue

    try:
        with open(f"{raw}/{file}", encoding="utf-8") as f:
            text = f.read()

        # 2. Excise the Gutenberg Boilerplate
        start = "*** START"
        end = "*** END"

        s = text.find(start)
        e = text.find(end)

        if s != -1 and e != -1:
            # Shift the start index to the end of the boilerplate line
            # so we don't include the "*** START OF..." text itself
            s_end_of_line = text.find('\n', s)
            if s_end_of_line != -1:
                text = text[s_end_of_line:e]
            else:
                text = text[s:e]

        # 3. Punctuation Normalization (CRITICAL for 40 novels)
        # Standardize archaic and 'smart' quotes to standard ascii
        text = re.sub(r'[“”]', '"', text)
        text = re.sub(r'[‘’]', "'", text)

        # Standardize em-dashes to spaced hyphens
        text = re.sub(r'—|--', ' - ', text)

        # 4. Whitespace Flattening
        text = re.sub(r"\n+", "\n", text)
        text = re.sub(r"\s+", " ", text)

        with open(f"{clean}/{file}", "w", encoding="utf-8") as f:
            f.write(text.strip())

        print(f"Purified: {file}")

    except Exception as err:
        print(f"Alas, an error thwarted the cleaning of {file}: {err}")

print("\nCleansing complete. The corpus is pristine and awaits extraction.")
```

```python
import nltk
from nltk.tokenize import sent_tokenize
import os
import json

# [Your excellent NLTK setup code remains exactly the same]
nltk_data_dir = '/root/nltk_data'
if nltk_data_dir not in nltk.data.path:
    nltk.data.path.append(nltk_data_dir)
os.makedirs(nltk_data_dir, exist_ok=True)

try:
    nltk.data.find('tokenizers/punkt')
    print("'punkt' NLTK resource found.")
except LookupError:
    print("'punkt' NLTK resource not found, attempting download...")
    nltk.download('punkt', download_dir=nltk_data_dir)

try:
    nltk.data.find('tokenizers/punkt_tab')
    print("'punkt_tab' NLTK resource found.")
except LookupError:
    print("'punkt_tab' NLTK resource not found, attempting download...")
    nltk.download('punkt_tab', download_dir=nltk_data_dir)

# --- The Enhanced Chunking Engine ---

clean = "VictorianGPT/cleaned"
chunks = []

# ARCHITECTURAL UPGRADES: Parameters for safe chunking
CHUNK_SIZE = 6    # Reduced to 6 sentences to respect embedding token limits
OVERLAP = 2       # 2 sentences will overlap between chunks to preserve context

if not os.path.exists(clean):
    print(f"Alas! Directory '{clean}' not found. Please ensure the archives are populated.")
else:
    for file_name in os.listdir(clean):
        file_path = os.path.join(clean, file_name)
        if os.path.isfile(file_path):
            with open(file_path, "r", encoding="utf8") as f:
                text = f.read()

            if text.strip():
                sentences = sent_tokenize(text)

                # The Sliding Window Approach
                # i increments by (CHUNK_SIZE - OVERLAP) to create interlocking context
                stride = CHUNK_SIZE - OVERLAP
                for i in range(0, len(sentences), stride):
                    block = " ".join(sentences[i:i + CHUNK_SIZE])

                    # QUALITY CONTROL: Ensure the block isn't empty and isn't obscenely long
                    if len(block.split()) > 10:
                        chunks.append(block)
            else:
                print(f"Warning: Skipping barren manuscript: {file_name}")

    print(f"\nSplendid! A total of {len(chunks)} interlocking chunks forged.")

# --- Disk Checkpointing ---
output_file = "VictorianGPT/rag_chunks_master.json"
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(chunks, f)

print(f"The corpus chunks have been securely entombed in {output_file}.")
```

```python
import pandas as pd

print(f"Binding the grand ledger from {len(chunks)} fragments...")

# 1. Forge the DataFrame
df = pd.DataFrame({"text": chunks})

# 2. Annihilate Redundancy
initial_count = len(df)
df = df.drop_duplicates(subset=["text"]).reset_index(drop=True)
final_count = len(df)

print(f"Purged {initial_count - final_count} duplicate fragments. {final_count} pristine chunks remain.")

# 3. The Modern Archive (Parquet)
# Note: You may need to run !pip install pyarrow in Colab if it is not installed
parquet_path = "VictorianGPT/chunks_master.parquet"
df.to_parquet(parquet_path, index=False)
print(f"Advanced Parquet archive sealed at: {parquet_path}")

# 4. The Fallback (Fortified CSV)
# If your downstream pipeline strictly requires a CSV, we use the escapechar
# parameter to prevent rogue commas and quotes from breaking the rows.
csv_path = "VictorianGPT/chunks_master.csv"
df.to_csv(csv_path, index=False, escapechar='\\')
print(f"Traditional CSV archive sealed at: {csv_path}\n")

# 5. Inspection
print("Behold the first entries of your new, massive corpus:")
print(df.head())
```
